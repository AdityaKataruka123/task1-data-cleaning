```python
# ================================
# DATA CLEANING TASK - NETFLIX DATASET
# ================================

# Import Libraries
import pandas as pd
import numpy as np

# Load Dataset
df = pd.read_csv("netflix_titles.csv")

# ================================
# DISPLAY BASIC INFORMATION
# ================================

print("========== FIRST 5 ROWS ==========")
print(df.head())

print("\n========== DATASET INFO ==========")
print(df.info())

print("\n========== DATASET SHAPE ==========")
print("Rows and Columns:", df.shape)

print("\n========== MISSING VALUES BEFORE CLEANING ==========")
print(df.isnull().sum())

print("\n========== DUPLICATE ROWS BEFORE CLEANING ==========")
print(df.duplicated().sum())

# ================================
# REMOVE DUPLICATES
# ================================

df.drop_duplicates(inplace=True)

# ================================
# HANDLE MISSING VALUES
# ================================

# Fill missing director values
df['director'] = df['director'].fillna('Unknown')

# Fill missing cast values
df['cast'] = df['cast'].fillna('Not Available')

# Fill missing country values
df['country'] = df['country'].fillna('Not Specified')

# Fill missing rating values
df['rating'] = df['rating'].fillna('Not Rated')

# Fill missing duration values
df['duration'] = df['duration'].fillna('Unknown')

# Remove rows where date_added is missing
df.dropna(subset=['date_added'], inplace=True)

# ================================
# STANDARDIZE TEXT FORMATTING
# ================================

# Remove extra spaces and capitalize properly
df['type'] = df['type'].str.strip().str.title()

df['country'] = df['country'].str.strip().str.title()

df['rating'] = df['rating'].str.strip().str.upper()

# ================================
# CONVERT DATE FORMAT
# ================================

df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Convert to dd-mm-yyyy format
df['date_added'] = df['date_added'].dt.strftime('%d-%m-%Y')

# ================================
# CLEAN COLUMN NAMES
# ================================

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

# ================================
# FIX DATA TYPES
# ================================

df['release_year'] = df['release_year'].astype(int)

# ================================
# FINAL DATA CHECKS
# ================================

print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

print("\n========== DUPLICATE ROWS AFTER CLEANING ==========")
print(df.duplicated().sum())

print("\n========== CLEANED DATASET INFO ==========")
print(df.info())

print("\n========== CLEANED FIRST 5 ROWS ==========")
print(df.head())

# ================================
# SAVE CLEANED DATASET
# ================================

df.to_csv("cleaned_netflix_titles.csv", index=False)

print("\n===================================")
print("DATA CLEANING COMPLETED SUCCESSFULLY")
print("Cleaned file saved as:")
print("cleaned_netflix_titles.csv")
print("===================================")
```
